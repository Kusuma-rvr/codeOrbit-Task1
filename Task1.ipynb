```python
import pandas as pd

df = pd.read_excel("sales_products_noisy_dataset_500_rows.xlsx")

print("Dataset loaded successfully!")
print(df.head())
```

```python
print("Rows and Columns:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicates removed")
print("New shape:", df.shape)
```

```python
df["Product"] = df["Product"].str.strip()
df["Category"] = df["Category"].str.strip()
df["Region"] = df["Region"].str.strip()
df["Customer_Type"] = df["Customer_Type"].str.strip()

print("Formatting cleaned successfully!")
```

```python
df["Product"] = df["Product"].replace({
    "laptop": "Laptop",
    "Smart Phone": "Smartphone",
    "Head phone": "Headphones"
})

print("Product names standardized!")
```

```python
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
).dt.date

print("Order dates cleaned successfully!")
```

```python
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Unit_Price"] = df["Unit_Price"].fillna(df["Unit_Price"].median())
df["Sales_Amount"] = df["Sales_Amount"].fillna(df["Sales_Amount"].median())
df["Discount"] = df["Discount"].fillna(df["Discount"].median())

print("Numeric missing values handled!")
```

```python
df["Product"] = df["Product"].fillna("Unknown")
df["Category"] = df["Category"].fillna("Unknown")
df["Region"] = df["Region"].fillna("Unknown")
df["Customer_Type"] = df["Customer_Type"].fillna("Unknown")

print("Text missing values handled!")
```

```python
print("Missing Values After Cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows After Cleaning:")
print(df.duplicated().sum())
```

```python
print("Cleaned Dataset:")
print(df.head(10))
```

```python
print("Final Rows and Columns:")
print(df.shape)
```

```python
df.to_excel("Cleaned_Sales_Products_Dataset.xlsx", index=False)

print("Cleaned dataset saved successfully!")
```
